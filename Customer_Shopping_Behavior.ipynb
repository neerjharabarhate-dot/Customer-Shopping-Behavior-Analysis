```python
import pandas as pd

df=pd.read_csv("customer_shopping_behavior.csv")

print("\n Head: ", df.head(5))

#------------------------ Data Checking----------------------
```

```python
print("\n Columns: ", df.columns)
```

```python
print("\n Empty Values: ", df.isnull().sum())
```

```python
print("\nInfo:")
df.info()
```

```python
print("\nShape:")
print(df.shape)
```

```python
# Check Duplicate Rows: 
print("\n Duplicate Rows: ", df.duplicated().sum())

# Check Unique Values: 
print("\n Unique Values: ")
for col in df.columns:
    print(col,":",df[col].unique())
```

```python
# Check rows contains same values: 

df[['discount_applied','promo_code_used']].head(5)
print((df["discount_applied"] == df["promo_code_used"]).all())
df=df.drop('promo_code_used',axis=1)
df.columns
```

```python
#------------------------ Data Cleaning----------------------

# Fixing Missing Values:
# df[df["Review Rating"].isnull()]

df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
df["Review Rating"].isnull().sum()
```

```python
#Standardizing Column Names: 

df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
# removing unnecessary column that were being created while cleaning the data
# df = df.drop(columns=["age_group"])
print(df.columns.tolist())
```

```python
#------------------------ Data Transfromation----------------------

# Creating Age Group:
bins=[17,25,35,45,60,100]
labels=[
    "Young Adult",
    "Adult",
    "Middle-Aged Adult",
    "Older Adult",
    "Senior"
]
df["age_grp"]=pd.cut(
    df["age"],
    bins=bins,
    labels=labels
)
df[['age','age_grp']]
```

```python
# Create Purchase Amount Category
bins=[0,30,60,100]
labels=[
    "low",
    "medium",
    "high"
]
df["purchase_amount_category"]=pd.cut(
    df["purchase_amount"],
    bins=bins,
    labels=labels
)
df[['purchase_amount','purchase_amount_category']].head(10)
```

```python
# Loyality Segment
bins=[0,10,30,50]
labels=[
    "low loyality",
    "medium loyality",
    "high loyality"
]
df['loyality_segment']=pd.cut(
    df['previous_purchases'],
    bins=bins,
    labels=labels
)
df[['previous_purchases','loyality_segment']].head(10)
```

```python
# Create a new column purchase_freq_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```
